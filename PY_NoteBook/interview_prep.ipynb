Class : is a collection of similar objects


```python
class car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        return f"{self.year} {self.make} {self.model}"


car1 = car("Toyota", "Camry", 2020)
print(car1.display_info())
```

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"My name is {self.name}"


student1 = Student("Rajeev", 23)

print(student1.introduce())
```

__init__ : is an initializer method that runs when an object is created

```python
class Car:

    def __init__(self, brand):
        self.brand = brand

car = Car("Toyota")
print(car.brand)    #when car("Toyata") is created, __init__ initialize
                    # the object with the brand attribute set to "Toyota"
```
