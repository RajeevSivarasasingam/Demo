Class : is a collection of similar objects


```python
class car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        return f"{self.year} {self.make} {self.model}"


car1 = car("Toyota", "Camry", 2020)
print(car1.display_info())
```

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"My name is {self.name}"


student1 = Student("Rajeev", 23)

print(student1.introduce())
```

__init__ : is an initializer method that runs when an object is created

```python
class Car:

    def __init__(self, brand):
        self.brand = brand

car = Car("Toyota")
print(car.brand)    #when car("Toyata") is created, __init__ initialize
                    # the object with the brand attribute set to "Toyota"
```

Inheritance: allow one class to reuse or extend another class

```python
class Animal:

    def speak(self):
        print("Animal sound")


class Dog(Animal):

    def bark(self):
        print("Woof")


dog = Dog()

dog.speak()
dog.bark()
```

```python
def logger(func):
    def wrapper():
        print("Function called")
        return func()

    return wrapper


@logger
def hello():
    print("Hello")


hello()
```

```python
def numbers():
    yield 1
    yield 2
    yield 3

for n in numbers():
    print(n)
```

```python
#Reverse String
def reverse_string(s):
    return s[::-1]

print(reverse_string("Hello, World!"))

#check palindrome
def pal(p):
    if(p==p[::-1]):
        print(p+": It is palindrome!!")
    else:
        print(p+": It's not palindrome text")

pal("madam")
pal("appa")
pal("pencil")
```

```python
numbers = [10, 45, 20, 80, 30]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print(largest)
print(max(numbers))
```

```python
# count characters
from itertools import count


n=input("Enter a word: ")
count ={}
for char in n:
    count[char] = count.get(char,0)+1

print(count)
```
